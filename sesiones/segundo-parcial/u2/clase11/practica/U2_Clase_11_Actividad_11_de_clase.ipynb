# Actividad 11
Obtenga la cotización diaria de cierre del Merval cuyo symbol= ^MERV para el período 1/1/2025 al 
1/12/2025 mediante la API de Yahoo Finance. Conviértalo a DataFrame. Restee el índice. Renombre 
las columnas como Fecha y Precio

```python
import yfinance as yf
import pandas as pd

# Descargamos las cotizaciones de cierre
merval = yf.download("^MERV", start="2025-01-01", end="2025-12-01")

# Nos quedamos solo con el precio de cierre
df = merval[["Close"]].reset_index()
df.columns = ["Fecha", "Precio"]

print(df.head())
print(f"\nn = {len(df)} días hábiles")
```

  Escriba un programa para resolver cada punto que muestre por pantalla la respuesta completa:  
  A) ¿Cuál es el la cotización promedio del Merval en el período?
  

```python
# A
print(f"Promedio: {df['Precio'].mean():.2f} pesos")
```

B) ¿En qué día la cotización alcanza el 50% y cuál es el valor?

```python
# B
p_50 = df["Precio"].quantile(0.5)
print(f"Valor donde alcanza el 50% de los datos: {p_50:.2f} pesos")
print(f"Fecha donde se alcanza el 50% de los datos: {df[df['Precio'] <= p_50]['Fecha'].iloc[-1].date()}")
```

C) ¿En qué día la cotización alcanza el 75% y cuál es el valor?

```python
# C
p_75 = df["Precio"].quantile(0.75)
print(f"Valor donde alcanza el 75% de los datos: {p_75:.2f} pesos")
print(f"Fecha donde se alcanza el 75% de los datos: {df[df['Precio'] <= p_75]['Fecha'].iloc[-1].date()}")
```

D) ¿En qué día la cotización es mínima para el 25% superior de las cotizaciones y cuál es el valor?

```python
# D
p_25 = df["Precio"].quantile(0.25)
print(f"Valor donde alcanza el 25% de los datos: {p_25:.2f} pesos")
print(f"Fecha donde se alcanza el 25% de los datos: {df[df['Precio'] <= p_25]['Fecha'].iloc[-1].date()}")
```

E) ¿En qué día la cotización es máxima para el 25% de las cotizaciones y cuál es el valor?

```python
q1 = df["Precio"].quantile(0.25)
bot25 = df[df["Precio"] <= q1]
idx_max = bot25["Precio"].idxmax()
print(f"Q1 = ${q1:,.2f}  (umbral del 25% inferior)")
print(f"Maximo del 25% inferior: ${bot25.loc[idx_max, 'Precio']:,.2f}")
print(f"Fecha: {bot25.loc[idx_max, 'Fecha'].date()}")
```

F) ¿Entre qué valores se encuentran las cotización del Merval?

```python
print(f"Las cotizaciones se encuentran entre ${df['Precio'].min():,.2f} y ${df['Precio'].max():,.2f}")
```

G) ¿Entre qué días se concentra el 50% de las cotizaciones y cuál es el valor acumulado?

```python
q1 = df["Precio"].quantile(0.25)
q3 = df["Precio"].quantile(0.75)
rango_iqr = df[(df["Precio"] >= q1) & (df["Precio"] <= q3)]
print("El 50% central se concentra entre:")
print(f"  Desde: {rango_iqr['Fecha'].min().date()}  (Q1 = ${q1:,.2f})")
print(f"  Hasta: {rango_iqr['Fecha'].max().date()}  (Q3 = ${q3:,.2f})")
print(f"  IQR   = ${q3 - q1:,.2f}")
print(f"  Valor acumulado: ${rango_iqr['Precio'].sum():,.2f}")
```

H) ¿Cuál es la variabilidad de la cotización en el período dado?

```python
print(f"Varianza: {df['Precio'].var():,.2f}")
```

I) ¿Cuál es la variabilidad de la cotización expresada en la misma unidad de medida que la cotización en el 
período dado?

```python
print(f"Desvio estandar: ${df['Precio'].std():,.2f}")
```

J) ¿Existe evidencia suficiente para afirmar que la cotización media del Merval es menor a $ 2.500.000 con un 
95% de confianza?

```python
from scipy import stats
t_stat, p_valor = stats.ttest_1samp(df["Precio"].dropna(), popmean=2_500_000, alternative='less')
alpha = 0.05
print("H0: media >= 2500000   |   H1: media < 2500000")
print(f"Estadistico t : {t_stat:.4f}")
print(f"p-valor       : {p_valor:.4f}")
print(f"alfa          : {alpha}")
print()
if p_valor < alpha:
    print("RECHAZAMOS H0: hay evidencia suficiente con 95% de confianza")
else:
    print("NO rechazamos H0: no hay evidencia suficiente")
```
