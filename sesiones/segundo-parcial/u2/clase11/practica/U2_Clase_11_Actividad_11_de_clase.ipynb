J) ¿Existe evidencia suficiente para afirmar que la cotización media del Merval es menor a $ 2.500.000 con un 
95% de confianza?

````python
from scipy import stats
t_stat, p_valor = stats.ttest_1samp(df["Precio"].dropna(), popmean=2_500_000, alternative='less')
alpha = 0.05
print("H0: media >= 2500000   |   H1: media < 2500000")
print(f"Estadistico t : {t_stat:.4f}")
print(f"p-valor       : {p_valor:.4f}")
print(f"alfa          : {alpha}")
print()
if p_valor < alpha:
    print("RECHAZAMOS H0: hay evidencia suficiente con 95% de confianza")
else:
    print("NO rechazamos H0: no hay evidencia suficiente")
````

## Inciso J — Test de hipótesis para la media (varianza desconocida)

### Planteo del test

Se quiere determinar si la media del Merval es **menor** a 2.500.000 con un nivel de significancia del 5%.

$$H_0: \mu \geq 2.500.000 \qquad H_1: \mu < 2.500.000 \qquad \alpha = 0{,}05$$

---

### Estadístico t de Student

Cuando la varianza poblacional es **desconocida**, se utiliza la distribución $t$ con $n-1$ grados de libertad:

$$t_{obs} = \frac{\bar{x} - \mu_0}{S / \sqrt{n}}$$

| Símbolo | Descripción |
|---|---|
| $\bar{x}$ | Media muestral |
| $\mu_0$ | Media hipotética bajo $H_0$ (en este caso: 2.500.000) |
| $S$ | Desvío estándar muestral |
| $n$ | Tamaño de la muestra |

---

### Criterios de decisión (hipótesis unilateral izquierda: $H_1: \mu < \mu_0$)

**Por región crítica:**

$$\text{Rechazar } H_0 \text{ si } t_{obs} < t_{\alpha;\, n-1}$$

**Por p-valor:**

$$p\text{-valor} = P(T_{n-1} \leq t_{obs}) = F_t(t_{obs};\, n-1)$$

$$\text{Rechazar } H_0 \text{ si } p\text{-valor} < \alpha$$

---

### Comparación: enfoque del profesor vs. `scipy.stats.ttest_1samp`

| Paso | Función del profesor (`TH_MEDIA_VARDESCON`) | `scipy.stats.ttest_1samp` |
|---|---|---|
| Estadístico $t$ | `tobs = (media_muestral - mu) / (S / sqrt(n))` | calculado internamente |
| Valor crítico | `talfa = ss.t.ppf(alfa, n-1)` | no lo devuelve directamente |
| p-valor | `p_valor = ss.t.cdf(tobs, n-1)` | devuelto como segundo elemento |
| Decisión | `if tobs < talfa` / `if p_valor < alfa` | `if p_valor < alfa` |
| Hipótesis alternativa | parámetro `simbolo_H1='<'` | parámetro `alternative='less'` |
| Resultado | texto interpretativo completo | solo `(t_stat, p_valor)` |

---

### Parámetros de la función del profesor

```python
TH_MEDIA_VARDESCON(media_muestral, mu, S, n, alfa, simbolo_H1)
```

| Parámetro | Tipo | Descripción |
|---|---|---|
| `media_muestral` | `float` | Media calculada sobre la muestra ($\bar{x}$) |
| `mu` | `float` | Valor de la media poblacional bajo $H_0$ ($\mu_0$) |
| `S` | `float` | Desvío estándar muestral ($S$) |
| `n` | `int` | Cantidad de observaciones de la muestra |
| `alfa` | `float` | Nivel de significancia ($\alpha$), típicamente `0.05` |
| `simbolo_H1` | `str` | Dirección de $H_1$: `'<'` (izquierda), `'>'` (derecha), `'!='` (bilateral) |

---

### Equivalencia en código

```python
# Enfoque del profesor (manual, paso a paso)
TH_MEDIA_VARDESCON(
    media_muestral = df["Precio"].mean(),
    mu             = 2_500_000,
    S              = df["Precio"].std(),
    n              = len(df["Precio"].dropna()),
    alfa           = 0.05,
    simbolo_H1     = '<'
)

# Equivalente con scipy (compacto)
t_stat, p_valor = stats.ttest_1samp(df["Precio"].dropna(), popmean=2_500_000, alternative='less')
```

Ambos aplican la misma fórmula y llegan al mismo $t_{obs}$ y p-valor. La diferencia es que la función del profesor también compara contra el valor crítico $t_{\alpha;\,n-1}$ y devuelve un texto de interpretación, mientras que `ttest_1samp` devuelve solo los números y deja la interpretación al programador.
