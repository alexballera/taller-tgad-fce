
## **Ejercicio 2**

Con los datos descargados elabore  un tablero estratégico que calcule y visualice los siguientes KPIs mediante gráficos indicadores, indicando con colores el nivel de cumplimiento de cada meta:
| KPI | Meta |
|-----|------|
| Ventas totales | $2.000.000 |
| Ganancia total | $300.000 |
| Cantidad de órdenes | 5.000 |
| Descuento promedio | máximo 20% |

### 2.1 KPIs individuales

Primero calculamos y visualizamos cada KPI por separado. Después los combinamos en el tablero.

#### KPI 1: Ventas totales


```python
# representa la suma total de ventas del período analizado
ventas_totales = round(DATA['Sales'].sum(), 2)  
print(f"Ventas totales: ${ventas_totales}")
```

```python
# Creamos el indicador 
kpi_ventas = go.Indicator(
    mode="gauge+number+delta",              
    value=ventas_totales,                   
    title={"text": "Ventas", "font": {"size": 18}},
    delta={"reference": 2000000},          
    number={"prefix": "$", "font": {"size": 40}},
    gauge={
        "axis": {"range": [0, 3000000]},    
        "bar": {"color": "#193853", "thickness": 0.3},
        "steps": [
            {"range": [0, 1000000],       "color": "#da5757"},  
            {"range": [1000000, 2000000], "color": "#d7a839"},  
            {"range": [2000000, 3000000], "color": "#6da86d"}   
        ],
        "threshold": {"line": {"color": "white", "width": 4}, "value": 2000000}  
    }
)
go.Figure(kpi_ventas).show()
```

#### KPI 2: Ganancia total


```python
# representa suma total de ganancias
ganancia_total = round(DATA['Profit'].sum(), 2)  
print(f"Ganancia total: ${ganancia_total}")
```

```python
# Creamos el indicador 
kpi_ganancia = go.Indicator(
    mode="gauge+number+delta",
    value=ganancia_total,                                       
    title={"text": "Ganancia", "font": {"size": 18}},
    delta={"reference": 300000},                                
    number={"prefix": "$", "font": {"size": 40}},
    gauge={
        "axis": {"range": [0, 500000]},
        "bar": {"color": "#193853", "thickness": 0.3},
        "steps": [
            {"range": [0, 150000],      "color": "#da5757"},
            {"range": [150000, 300000], "color": "#d7a839"},
            {"range": [300000, 500000], "color": "#6da86d"}
        ],
        "threshold": {"line": {"color": "white", "width": 4}, "value": 300000}    
    }
)
go.Figure(kpi_ganancia).show()
```

#### KPI 3 : Cantidad de órdenes


```python
# representa la cantidad de órdenes únicas
cantidad_ordenes = DATA['Order ID'].nunique()  
print(f"Cantidad de órdenes: {cantidad_ordenes}")
```

```python
# Creamos el indicador 
kpi_ordenes = go.Indicator(
    mode="gauge+number+delta",
    value=cantidad_ordenes,
    title={"text": "Órdenes", "font": {"size": 18}},
    delta={"reference": 5000},                    
    number={"font": {"size": 40}},
    gauge={
        "axis": {"range": [0, 7000]},
        "bar": {"color": "#193853", "thickness": 0.3},
        "steps": [
            {"range": [0, 2500],    "color": "#da5757"},
            {"range": [2500, 5000], "color": "#d7a839"},
            {"range": [5000, 7000], "color": "#6da86d"}
        ],
        "threshold": {"line": {"color": "white", "width": 4}, "value": 5000}  
    }
)
go.Figure(kpi_ordenes).show()
```

#### KPI 4: Descuento promedio


```python
# representa el descuento promedio otorgado
descuento_prom = round(DATA['Discount'].mean() * 100, 1)  
print(f"Descuento promedio: {descuento_prom}%")
```

```python
# Creamos el indicador
kpi_descuento = go.Indicator(
    mode="gauge+number+delta",
    value=descuento_prom,
    title={"text": "Descuento Promedio", "font": {"size": 18}},
    delta={
        "reference": 20,
        "increasing": {"color": "red"},     
        "decreasing": {"color": "green"}    
    },
    number={"suffix": "%", "font": {"size": 40}},
    gauge={
        "axis": {"range": [0, 50]},
        "bar": {"color": "#193853", "thickness": 0.3},
        "steps": [
            {"range": [0, 20],  "color": "#6da86d"},   
            {"range": [20, 35], "color": "#d7a839"},   
            {"range": [35, 50], "color": "#da5757"}    
        ],
        "threshold": {"line": {"color": "white", "width": 4}, "value": 20}
    }
)
go.Figure(kpi_descuento).show()
```

### 2.2. Tablero integrado

Combinamos los gráficos en un único tablero con `make_subplots`.


```python
# grilla 2x2: row=2, cols=2
fig = make_subplots(
    rows=2, cols=2,
    specs=[
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "indicator"}, {"type": "indicator"}]
    ]
)

# llamamos a las variables ya definidas y los ubicamos en celdas:
fig.add_trace(kpi_ventas,    row=1, col=1)
fig.add_trace(kpi_ganancia,  row=1, col=2)
fig.add_trace(kpi_ordenes,   row=2, col=1)
fig.add_trace(kpi_descuento, row=2, col=2)

# diseño layout
fig.update_layout(
    title_text="TABLERO ESTRATÉGICO",
    title_x=0.5,
    title_font={"size": 24},
    template="plotly_white",
    height=700                    
)
fig.show()
```

## **Ejercicio 3** 

La misma empresa quiere analizar sus ventas y ganancias entre 2014 y 2017. Elaborá un tablero analítico que combine:

- Evolución de ventas mensuales → gráfico de líneas
- Ventas por categoría → gráfico de barras
- Participación por segmento → gráfico de torta
- Distribución de ganancias por región → boxplot


###  3.1.Gráficos individuales

Primero calculamos y visualizamos cada gráfico por separado. Después los combinamos en el tablero.

#### Gráfico 1: Evolución de ventas por mes


```python
# convertimos la fecha a datetime
DATA['Order Date'] = pd.to_datetime(DATA['Order Date'])
```

```python
# agrupamos por mes y sumamos 
ventas_mes = DATA.groupby(DATA['Order Date'].dt.to_period('M'))['Sales'].sum().reset_index()
```

```python
# convertimos a string 
ventas_mes['Order Date'] = ventas_mes['Order Date'].astype(str)  
ventas_mes.head(3)
```

```python
# Creamos el gráfico
trace_ventas_mes = go.Scatter(
    x=ventas_mes['Order Date'],   
    y=ventas_mes['Sales'],        
    mode='lines+markers',         
    line={"color": "#636EFA"},
    name="Ventas"
)
go.Figure(trace_ventas_mes).show()
```


#### Gráfico 2:Ventas por categoría


```python
# agrupo ventas por categoría y renombro
ventas_categoria = DATA.groupby('Category')['Sales'].sum().reset_index()
ventas_categoria.columns = ['categoria', 'ventas']
```

```python
# ordeno de mayor a menor
ventas_categoria = ventas_categoria.sort_values('ventas', ascending=False)
ventas_categoria
```

```python
# Creamos el gráfico
trace_categorias = go.Bar(
    x=ventas_categoria['categoria'],   
    y=ventas_categoria['ventas'],      
    marker_color=['#636EFA', '#EF553B', '#00CC96'],   
    text=round(ventas_categoria['ventas'], 0),               
    textposition='auto'
)
go.Figure(trace_categorias).show()
```


#### Gráfico 3: Ventas por segmento


```python
# agrupo por segmento 
ventas_segmento = DATA.groupby('Segment')['Sales'].sum().reset_index()
ventas_segmento.columns = ['segmento', 'ventas']
ventas_segmento
```

```python
# Creamos el gráfico
trace_segmento = go.Pie(
    labels=ventas_segmento['segmento'],  
    values=ventas_segmento['ventas'],     
    hole=0.3                              
)
go.Figure(trace_segmento).show()
```


#### Gráfico 4: Ganancias por región


```python
# Creamos el gráfico
trace_boxplot = go.Box(
    x=DATA['Region'],   
    y=DATA['Profit']    
)
go.Figure(trace_boxplot).show()
```

### 3.2.Tablero analítico 

Combinamos los gráficos en un único tablero con `make_subplots`.


```python
# grilla 2x2: row=2, cols=2

fig = make_subplots(
    rows=2, cols=2,
    specs=[                                         
        [{"type": "xy"},     {"type": "xy"}],         
        [{"type": "domain"}, {"type": "xy"}]         
    ],
    subplot_titles=(
        "Tendencia de ventas por Mes y Año",
        "Ventas por Categoría",
        "Ventas por Segmento",
        "Distribución de Ganancias por Región"
    )
)

# llamamos a las variables ya definidas
fig.add_trace(trace_ventas_mes, row=1, col=1)
fig.add_trace(trace_categorias, row=1, col=2)
fig.add_trace(trace_segmento,   row=2, col=1)
fig.add_trace(trace_boxplot,    row=2, col=2)

# diseño layout
fig.update_layout(
    title_text="Tablero analítico 2014 a 2017",
    title_x=0.5,
    template="plotly_white",
    height=700,
    showlegend=False
)
fig.show()
```
