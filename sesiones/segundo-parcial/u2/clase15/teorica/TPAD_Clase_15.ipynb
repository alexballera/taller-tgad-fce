### 📊 Gráfico Indicador (`go.Indicator`)

El **gráfico indicador** (o *gauge chart*) es ideal para mostrar **un único valor clave en contexto**: cuánto representa respecto a un rango posible y cómo varió respecto a una referencia.

**Componentes principales:**
| Componente | Descripción |
|---|---|
| `value` | El valor actual a mostrar (ej. precio de cierre de julio) |
| `delta` | Variación respecto a una referencia (ej. precio del mes anterior) — muestra positivo/negativo en color |
| `gauge` | Medidor tipo velocímetro con zonas de color que definen rangos de interpretación |
| `mode` | Combinación de `gauge`, `number` y/o `delta` según lo que se quiera mostrar |

**Cuándo usarlo:**
- KPIs y tableros de control (*dashboards*)
- Indicadores financieros: precio actual vs. objetivo, avance de presupuesto, etc.
- Variables con un rango natural definido (0–100%, 0–máximo histórico)

**Interpretación:** En el ejemplo, el precio de cierre de Google en julio (131,6 USD) se muestra con **delta positivo** respecto a junio (118,7 USD), ubicado en la zona gris del velocímetro. La línea roja de umbral marca el máximo del rango (150 USD). La variante *bullet* (segunda celda) presenta el mismo dato en formato horizontal, más compacto para comparar múltiples indicadores. La tercer variante muestra **solo el delta** cuando el foco está en la variación, no en el valor absoluto.

### 📊 Gráfico Bullet

El **gráfico bullet** es una alternativa mejorada al velocímetro, diseñada por Stephen Few para dashboards compactos. Permite mostrar **un valor real comparado contra bandas de referencia** (metas, rangos cualitativos) en muy poco espacio.

**Componentes:**
| Elemento | Descripción |
|---|---|
| Barra principal (oscura/negra) | Valor actual de la métrica (ej. cantidad de empresas por sector) |
| Bandas de fondo (degradé) | Zonas cualitativas: bajo / medio / bueno / excelente |
| Marcador de referencia | Línea o punto que indica un objetivo o umbral |

**Ventajas sobre el gauge (velocímetro):**
- Ocupa mucho menos espacio → permite mostrar muchos KPIs en una misma pantalla
- Soporta fácilmente comparaciones entre múltiples categorías en formato apilado vertical
- Más preciso visualmente: la longitud de la barra es más fácil de comparar que el ángulo de un arco

**Interpretación:** En el ejemplo con sectores del S&P 500:
- Cada barra horizontal muestra la **cantidad de empresas** de ese sector
- Las bandas de color (verde degradé) marcan los rangos: menos de 20, 20–40, 40–60, etc.
- El sector `Financials` y `Industrials` son los más numerosos, superando los 60 componentes
- La versión `ff.create_bullet` de Plotly automatiza la creación para múltiples categorías a partir de un DataFrame con formato JSON

**Cuándo usarlo:** Performance de vendedores vs. cuota, indicadores operativos (tasa de error vs. benchmark), comparaciones sectoriales donde el foco está en el ranking relativo.


### **Gráfico Cascada**

````python
#obtención de datos
import yfinance as yf
VISA = pd.DataFrame(yf.download('V', start='2023-07-21', end='2023-08-01')).reset_index()
VISA.columns=VISA.columns.droplevel(1)
VISA
````

````python
VISA_1=VISA[['Date', 'Open', 'Close']]
VISA_1
````

````python
VISA_1.dtypes
````

````python
#cálculo de variación absoluta y porcentual
VISA_1['Var_Diaria']=round(VISA['Close']-VISA['Open'],1)
VISA_1['Var%_Diaria']=round(((VISA['Close']-VISA['Open'])/VISA['Open'])*100, 1)
VISA_1
````

````python
#cáluclo de suma acumulada
VISA_1["Cumulative"] = VISA_1['Var_Diaria'].cumsum()
VISA_1["Cumulative"] = [cum-val if val<0 else cum for cum, val in VISA_1[["Cumulative", "Var_Diaria"]].values]

## Bar Colors
VISA_1["Color"] = ["darkgreen" if val>0 else "darkred" if val<0 else "dodgerblue" for val in VISA_1["Var_Diaria"]]

VISA_1
````

````python
bottom = [0,]
height = [VISA_1['Var_Diaria'][0],]

for i, val in enumerate(VISA_1['Var_Diaria'][1:], start=1):
    if val==0:
        bottom.append(0)
        height.append(VISA_1["Cumulative"][i])
    elif val > 0:
        if VISA_1['Var_Diaria'][i-1] >=0:
            bottom.append(VISA_1["Cumulative"][i-1])
        else:
            bottom.append(bottom[i-1])
        height.append(val)
    elif val < 0:
        if VISA_1['Var_Diaria'][i-1] >=0:
            bottom.append(VISA_1["Cumulative"][i-1]+val)
        else:
            bottom.append(bottom[i-1]+val)
        height.append(-val)

VISA_1["Bottom"] = bottom
VISA_1["Height"] = height

VISA_1
````

````python
fig, ax = plt.subplots(figsize=(15,8))

plt.bar(x=VISA_1.index, height=VISA_1["Height"], bottom=VISA_1["Bottom"], color=VISA_1["Color"])

plt.xticks(VISA_1.index, VISA_1["Date"].dt.strftime('%Y-%m-%d'), fontdict=dict(fontsize=9))
plt.yticks(range(-2, 6, 1), ["$ {:,}".format(val) for val in range(-2, 6, 1)],fontdict=dict(fontsize=14))

for idx in range(len(VISA_1)):
    plt.text(x=VISA_1.index[idx], y=VISA_1["Cumulative"][idx],
             s="{:,}%".format(VISA_1["Var%_Diaria"][idx] if VISA_1["Var_Diaria"][idx]!=0 else VISA_1["Cumulative"][idx]),
             ha="right", va="bottom", fontdict=dict(fontsize=16))
    plt.text(x=VISA_1.index[idx], y=VISA_1["Cumulative"][idx],
             s="${:,}".format(VISA_1["Var_Diaria"][idx] if VISA_1["Var_Diaria"][idx]!=0 else VISA_1["Cumulative"][idx]),
             ha="left", va="bottom", fontdict=dict(fontsize=16))

plt.xlabel("Date", fontdict=dict(fontsize=16, fontweight="bold"))
plt.ylabel("Var_Diaria", fontdict=dict(fontsize=16, fontweight="bold"))
plt.title("Variación diaria", loc="left", pad=10, fontdict=dict(fontsize=20, fontweight="bold"))

ax.spines[[ "top", "right"]].set_visible(False)

plt.show()
````

### 📊 Gráfico de Cascada (*Waterfall*)

El **gráfico de cascada** muestra cómo un valor inicial se transforma en un valor final a través de una serie de aumentos y disminuciones, visualizando **la contribución de cada paso al resultado acumulado**.

**Estructura:**
- Cada barra comienza donde terminó la anterior (efecto "flotante")
- Barras **verdes** → variación positiva (ganancia, suba)
- Barras **rojas** → variación negativa (pérdida, baja)
- Barra total → valor acumulado final (barra desde cero)

**Interpretación del ejemplo:** La variación diaria del precio de VISA (julio 2023) muestra:
- Los días de **suba** aparecen en verde, los de **baja** en rojo
- Las etiquetas muestran simultáneamente la variación **absoluta** (en USD) y **porcentual** (%)
- El acumulado permite ver si la semana fue positiva o negativa en términos netos

**Conceptos clave de la implementación:**
| Concepto | Función |
|---|---|
| `cumsum()` | Suma acumulada para calcular el "piso" de cada barra |
| `bottom` + `height` | Posición vertical y altura de cada barra flotante |
| `go.Waterfall` | Versión interactiva que calcula automáticamente los "pisos" con `measure='relative'` |
| `measure='total'` | Marca la última barra como totalizadora (desde cero) |

**Cuándo usarlo:** Estados financieros (bridge de EBITDA), análisis de variaciones presupuestarias, composición de ganancias/pérdidas en períodos de tiempo, flujos de caja.

### 📊 Gráfico de Radar (*Spider Chart*)

El **gráfico de radar** (o gráfico araña / *spider chart*) muestra múltiples variables cuantitativas para una o más entidades en un sistema de ejes radiales, permitiendo ver **el perfil o forma** de cada entidad y compararlas visualmente.

**Estructura:**
- Cada eje representa una variable (en el ejemplo: `Price`, `52 Week Low`, `52 Week High`)
- El área encerrada por los vértices describe el **perfil** del sector
- Cuanto mayor el área, mayor son los valores en todas las dimensiones

**Interpretación del ejemplo:**
- Se comparan sectores del S&P 500 en tres métricas de precio
- Un sector con gran diferencia entre `52 Week Low` y `52 Week High` tiene alta **volatilidad** anual
- La forma asimétrica del polígono indica que el sector se destaca en algunas dimensiones más que en otras
- Con `px.line_polar` se obtiene una versión rápida; con `go.Scatterpolar` se tiene control total sobre colores, relleno y múltiples trazas superpuestas

**Limitaciones importantes:**
- Con más de 4–5 grupos el gráfico se vuelve difícil de leer por superposición
- Las variables deben estar en unidades comparables (o normalizadas) para que la forma tenga sentido
- El orden de los ejes afecta la percepción del área → con ejes reordenados, la misma figura puede parecer muy diferente

**Cuándo usarlo:** Comparación de perfiles (competidores, atletas, productos), evaluaciones multidimensionales con un número reducido de variables (3–8), análisis de fortalezas/debilidades.

### 📊 Gráfico de Puntos Conectados (*Dumbbell / Dot Plot*)

El **gráfico de puntos conectados** (también llamado *dumbbell chart* o *slope chart*) muestra la **variación de un valor entre dos momentos o condiciones** para múltiples categorías, usando puntos conectados por una línea o barra.

**Cuándo elegirlo sobre un gráfico de barras:**
- El foco está en el **cambio** entre dos estados, no en el valor absoluto
- Hay muchas categorías → las barras agrupadas se vuelven difíciles de leer
- Se quiere mostrar **dirección** (subió/bajó) y **magnitud** en una sola visual

**Componentes del ejemplo:**
| Elemento | Código Plotly |
|---|---|
| Punto azul claro | `go.Scatter(mode='markers')` — precio en junio |
| Punto azul oscuro | `go.Scatter(mode='markers')` — precio en julio |
| Barra conectora | `go.Bar(orientation='h', base=precio_junio)` — diferencia entre los dos valores |
| Etiqueta de variación | `go.Scatter(mode='text')` — muestra el % de cambio |

**Interpretación:** 
- Activos cuya barra conectora apunta a la **derecha** (bar positiva) subieron de precio en julio vs. junio
- El activo `T_MSFT` muestra la mayor suba proporcional
- `E_SHEL` muestra una baja: su punto de julio queda a la izquierda del de junio
- La segunda versión con `fig.add_annotation` agrega etiquetas manuales más precisas para cada barra, mejorando la legibilidad

**Limitaciones:** No es apropiado para más de 2 puntos de comparación; para 3 o más períodos, el *slope chart* o el gráfico de líneas es más adecuado.

### 📊 Gráfico de Coordenadas Paralelas (*Parallel Coordinates*)

El **gráfico de coordenadas paralelas** es una técnica de visualización **multivariada** que muestra cada observación como una línea que atraviesa múltiples ejes verticales paralelos, uno por cada variable.

**Estructura:**
- Cada eje vertical representa una variable (en el ejemplo: precio en junio y en julio)
- Cada línea representa un activo financiero
- La posición donde la línea cruza cada eje indica el valor de esa variable para esa observación
- El color codifica una variable adicional (en el ejemplo, el índice del activo)

**Ventajas:**
- Permite visualizar patrones en **muchas variables simultáneamente** (algo imposible con scatter 2D)
- Se puede identificar **clusters** (líneas con trayectorias similares) y **outliers** (líneas con cruces atípicos)
- Con Plotly es **interactivo**: se puede filtrar arrastrando rangos sobre cada eje

**Interpretación del ejemplo:**
- Los activos cuyas líneas van de izquierda hacia abajo-derecha bajaron de precio de junio a julio
- Los activos cuyas líneas van hacia arriba-derecha subieron
- `T_MSFT` y `T_GOOG` son visibles como los que más subieron (cruzan el eje de julio en posición alta)
- `E_SHEL` tiene el precio más alto en ambos meses pero con leve baja

**Cuándo usarlo:** Análisis exploratorio multivariado, perfiles de clientes con muchas características, comparación de modelos o configuraciones con múltiples métricas, datos financieros con múltiples períodos o activos.

### 📊 Botones (`updatemenus`)

Los **botones** en Plotly permiten agregar interactividad sin necesidad de un servidor backend: el usuario puede cambiar lo que se muestra haciendo clic, sin re-ejecutar código Python.

**Tipos de botones (`method`):**
| Método | Efecto |
|---|---|
| `'update'` | Modifica tanto los datos como el layout (el más completo) |
| `'restyle'` | Solo modifica los datos de los trazos (más eficiente para cambiar series) |
| `'relayout'` | Solo modifica el layout (ejes, título, etc.) |
| `'animate'` | Activa animaciones predefinidas |

**Ejemplo 1 — Botón de visibilidad:** muestra/oculta trazos individuales (`visible: True/False`). Se usa para alternar entre `Close`, `Open`, `High`, `Low` del precio de Google. Cada botón define exactamente qué trazos son visibles y cuál es el título del gráfico.

**Ejemplo 2 — Botón de agregación:** usa `transforms` de Plotly para cambiar la función agregada (avg, median, min, max, etc.) sobre los datos **en el cliente**, sin re-ejecutar Python. Esto es posible gracias al objeto `transforms[0].aggregations[0].func` que vive en el JSON del gráfico.

**Clave de implementación:**
```python
fig.update_layout(updatemenus=[dict(
    buttons=[dict(label='Opción A', method='update',
                  args=[{'visible': [True, False]}, {'title': 'Gráfico A'}])],
    direction='down'  # menú desplegable
)])
```

**Cuándo usarlo:** Dashboards exportados como HTML estático, notebooks donde se quiere evitar widgets interactivos que requieren kernel activo, exploración rápida de múltiples vistas de los mismos datos.

### 📊 Filtro con `updatemenus` y tablas (`go.Table`)

El **filtro** en Plotly permite al usuario seleccionar un subconjunto de datos sin re-ejecutar código, modificando dinámicamente lo que muestra el gráfico o la tabla.

**Diferencia con el botón de visibilidad:**
- El botón de visibilidad muestra/oculta trazos que **ya están cargados**
- El filtro con `method='restyle'` **reemplaza los datos** del trazo con un subconjunto

**Patrones del ejemplo:**

**Filtro en gráfico de líneas:**
```python
buttons.append(dict(
    method='restyle',
    args=[{'y': [DATA_YAHOO[col]], 'x': [DATA_YAHOO.index]}]
))
```
Cada botón actualiza `y` y `x` del único trazo visible, mostrando solo el activo seleccionado.

**Filtro en tabla (`go.Table`):**
```python
args=[{'cells': {'values': DATA_YAHOO_M.loc[DATA_YAHOO_M['ACTIVO'].eq(c)].T.values}}]
```
El menú filtra las filas de la tabla reemplazando los valores de las celdas. Con `'All'` se restauran todos los datos.

**Doble filtro:** Al tener dos `updatemenus` con distintas posiciones `y`, se pueden aplicar filtros independientes por dos columnas diferentes (ej. por `ACTIVO` y por `Date`), lo que aumenta significativamente el poder exploratorio del dashboard sin requerir servidor.

**Ventaja clave:** Todo funciona en un HTML estático exportado — sin servidor, sin Python corriendo, solo JavaScript del lado del cliente.

### 📊 Controles Deslizantes y Animaciones (`animation_frame`)

Los **controles deslizantes** (*sliders*) y las **animaciones** permiten agregar una dimensión temporal (o cualquier variable discreta) a un gráfico estático, convirtiendo una visualización 2D en una exploración de **3 dimensiones** donde la tercera es el tiempo u otra variable de agrupación.

**`px.scatter` con `animation_frame`:**
```python
px.scatter(..., animation_frame='Año-Mes', animation_group='SECTOR',
           size='Volume', color='ACTIVO')
```
- `animation_frame`: la variable que define cada fotograma de la animación (ej. mes/año)
- `animation_group`: mantiene la identidad de cada punto entre fotogramas (evita que "salten")
- `size`: codifica una tercera variable numérica con el tamaño del punto (aquí el volumen operado)
- El botón ▶ reproduce la animación automáticamente; el slider permite ir fotograma a fotograma

**Interpretación del ejemplo:**
- Cada punto es un activo financiero en un mes determinado
- El eje X muestra el período, el Y el precio de cierre
- El tamaño del punto refleja el volumen operado: puntos grandes = mayor liquidez
- Al reproducir la animación se observa la **trayectoria temporal** de cada activo

**Diferencia entre slider y filtro:**
| | Filtro (`updatemenus`) | Slider / Animación |
|---|---|---|
| Navegación | Por categoría (botón) | Secuencial (frame a frame) |
| Propósito | Explorar subconjuntos | Ver evolución temporal |
| Animación automática | No | Sí (botón ▶) |

**Cuándo usarlo:** Series temporales con muchos períodos, evolución de indicadores a lo largo del tiempo, presentaciones donde se quiere "contar una historia" con los datos avanzando cronológicamente.
